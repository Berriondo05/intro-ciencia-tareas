## 2. Convertir a DataFrames

## 3. Cargar tablas a SQLite

```python
conn = sqlite3.connect(":memory:")

estudiantes.to_sql("estudiantes", conn, index=False, if_exists="replace")
cursos.to_sql("cursos", conn, index=False, if_exists="replace")
inscripciones.to_sql("inscripciones", conn, index=False, if_exists="replace")
pagos.to_sql("pagos", conn, index=False, if_exists="replace")

def run_sql(query):
    return pd.read_sql_query(query, conn)
```

Trae todos los estudiantes que viven en **Bogotá**.

### Punto 4. Agregación simple

Calcula:
- número total de cursos
- precio promedio
- precio mínimo
- precio máximo

Usa la tabla `cursos`.

**Entregable:**
- SQL
- pandas

```python
query_4 = """
SELECT COUNT(*) AS total_cursos,
       AVG(precio) AS precio_promedio,
       MIN(precio) AS precio_minimo,
       MAX(precio) AS precio_maximo
FROM cursos;
"""

(run_sql(query_4))
```

```python
cursos.agg(
    total_cursos=("curso_id", "count"),
    precio_promedio=("precio", "mean"),
    precio_minimo=("precio", "min"),
    precio_maximo=("precio", "max")
)
```

### Punto 5. GROUP BY

Cuenta cuántos cursos hay por `categoria`.

Ordena de mayor a menor cantidad.

**Entregable:**
- SQL
- pandas

```python
query_5 = """
SELECT categoria, COUNT(*) AS cantidad_cursos
FROM cursos
GROUP BY categoria
ORDER BY cantidad_cursos DESC;
"""

(run_sql(query_5))
```

```python
cursos.groupby("categoria").size().sort_values(ascending=False).reset_index(name="cantidad_cursos")
```

### Punto 6. JOIN entre tablas

Muestra:
- nombre del estudiante
- nombre del curso
- estado de la inscripción

Usa `estudiantes`, `inscripciones` y `cursos`.

**Entregable:**
- SQL
- pandas

```python
query_6 = """
SELECT estudiantes.nombre, cursos.nombre_curso, inscripciones.estado
FROM estudiantes
JOIN inscripciones
ON estudiantes.estudiante_id = inscripciones.estudiante_id
JOIN cursos
ON inscripciones.curso_id = cursos.curso_id;
"""

(run_sql(query_6))
```

```python
resultado = estudiantes.merge(inscripciones, on="estudiante_id").merge(cursos, on="curso_id")

resultado[["nombre", "nombre_curso", "estado"]]
```

### Punto 7. LEFT JOIN

Trae todos los estudiantes, incluso si no tuvieran inscripciones, mostrando:
- `estudiante_id`
- `nombre`
- `inscripcion_id`

**Entregable:**
- SQL
- pandas

> Pista: en este dataset puede que todos tengan al menos una inscripción, pero igual debes practicar el `LEFT JOIN`.

```python
query_7 = """
SELECT estudiantes.estudiante_id, estudiantes.nombre, inscripciones.inscripcion_id
FROM estudiantes
LEFT JOIN inscripciones
ON estudiantes.estudiante_id = inscripciones.estudiante_id;
"""

(run_sql(query_7))
```

```python
estudiantes.merge(inscripciones, on="estudiante_id", how="left")[["estudiante_id", "nombre", "inscripcion_id"]]
```

### Punto 8. HAVING

Calcula cuántas inscripciones tiene cada estudiante y muestra solo aquellos con **2 o más**.

Debes mostrar:
- `estudiante_id`
- `nombre`
- `n_inscripciones`

**Entregable:**
- SQL
- pandas

```python
query_8 = """
SELECT estudiantes.estudiante_id, estudiantes.nombre, COUNT(inscripciones.inscripcion_id) AS n_inscripciones
FROM estudiantes
JOIN inscripciones
ON estudiantes.estudiante_id = inscripciones.estudiante_id
GROUP BY estudiantes.estudiante_id, estudiantes.nombre
HAVING COUNT(inscripciones.inscripcion_id) >= 2;
"""

(run_sql(query_8))
```

```python
resultado = estudiantes.merge(inscripciones, on="estudiante_id")

resultado.groupby(["estudiante_id", "nombre"]).size().reset_index(name="n_inscripciones").query("n_inscripciones >= 2")
```

### Punto 9. CASE WHEN / segmentación

Clasifica los cursos según su precio:
- `Bajo` si el precio es menor a 120
- `Medio` si el precio está entre 120 y 180 inclusive
- `Alto` si el precio es mayor a 180

Debes mostrar:
- `nombre_curso`
- `precio`
- `segmento_precio`

**Entregable:**
- SQL
- pandas

```python
query_9 = """
SELECT nombre_curso, precio,
       CASE
           WHEN precio < 120 THEN 'Bajo'
           WHEN precio <= 180 THEN 'Medio'
           ELSE 'Alto'
       END AS segmento_precio
FROM cursos;
"""

(run_sql(query_9))
```

```python
cursos["segmento_precio"] = cursos["precio"].apply(
    lambda x: "Bajo" if x < 120 else "Medio" if x <= 180 else "Alto"
)

cursos[["nombre_curso", "precio", "segmento_precio"]]
```

### Punto 10. Ejercicio final integrador

Calcula el total pagado por categoría de curso.

Debes:
- unir `cursos`, `inscripciones` y `pagos`
- agrupar por `categoria`
- sumar `monto`
- ordenar de mayor a menor

**Entregable:**
- SQL
- pandas

```python
query_10 = """
SELECT cursos.categoria, SUM(pagos.monto) AS total_pagado
FROM cursos
JOIN inscripciones
ON cursos.curso_id = inscripciones.curso_id
JOIN pagos
ON inscripciones.inscripcion_id = pagos.inscripcion_id
GROUP BY cursos.categoria
ORDER BY total_pagado DESC;
"""

display(run_sql(query_10))
```

```python
resultado = cursos.merge(inscripciones, on="curso_id").merge(pagos, on="inscripcion_id")

resultado.groupby("categoria")["monto"].sum().sort_values(ascending=False).reset_index(name="total_pagado")
```
